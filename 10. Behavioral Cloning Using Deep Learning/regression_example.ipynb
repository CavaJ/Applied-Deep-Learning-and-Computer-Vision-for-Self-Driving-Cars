### Regression Example

```python
# import necessary packages
import numpy as np
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.optimizers import Adam
```

```python
# let's generate non-linear data
np.random.seed(0)
points = 500
X = np.linspace(-3, 3, points)
y = np.sin(X) + np.random.uniform(-0.5, 0.5, points)
plt.scatter(X, y)
```

```python
# define the neural network architecture using Keras
model = tf.keras.Sequential()
model.add(tf.keras.layers.Dense(50, activation='sigmoid', input_dim=1))
model.add(tf.keras.layers.Dense(30, activation='sigmoid'))
model.add(tf.keras.layers.Dense(1))
```

```python
# we will compile our model using the Adam optimizer. 
# our loss function will be MSE.
# 50 epochs will be used.
adam = Adam(lr=0.01)
model.compile(loss='mse', optimizer=adam)
model.fit(X, y, epochs=50)
```

```python
predictions = model.predict(X)
plt.scatter(X, y)
plt.plot(X, predictions, 'ro')
plt.show()
```
